```python
import numpy as np
import scipy.sparse as sp

import matplotlib.pyplot as plt

from dalia.backend.datastructures import DenseMatrix, SparseMatrix
```

```python
# Create the basic block matrices
# . DenseMatrix
shape = (3, 3)
data_00 = np.arange(1, shape[0] * shape[1] + 1).reshape(shape)
data_00 = data_00.astype(np.float64)
dmat_00 = DenseMatrix(data_00)

data_10 = np.ones(shape)
data_10 = data_10.astype(np.float64)
dmat_10 = DenseMatrix(data=data_10)

# . SparseMatrix
data_01 = np.random.rand(*shape)
density = 0.5
data_01[data_01 < density] = 0
data_01 = data_01.astype(np.float64)
sp_01 = sp.csr_matrix(data_01)
smat_01 = SparseMatrix(data=sp.csr_matrix(data_01))
```

```python
from dalia.backend.datastructures import BStructMatrix

bstruct = BStructMatrix([[dmat_00, smat_01], [dmat_10, None]])
print(bstruct)
```
